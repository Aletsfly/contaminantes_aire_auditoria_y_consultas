```python
import pandas as pd
import glob
import os
```

### Importación de 8 archivos csv y concatenación

```python
# 1. Definimos mi ruta a la carpeta que contiene mis archivos CSV.
#Ahora apuntamos directamente a la carpeta "consolidados".
ruta_carpeta = 'consolidados' 

# 2. Crea el patrón para encontrar todos los archivos .csv en esa carpeta.
patron_archivos = os.path.join(ruta_carpeta, '*.csv')

# 3. Usa glob para obtener una lista con las rutas de todos tus archivos CSV.
lista_archivos_csv = glob.glob(patron_archivos)
print(f"Archivos encontrados: {len(lista_archivos_csv)}")
print(lista_archivos_csv)
```

```python
if not lista_archivos_csv:
    print(f"\n⚠️ ¡Advertencia! No se encontraron archivos .csv en la carpeta '{ruta_carpeta_entrada}'.")
else:
    # 4. Lee y une los archivos en un solo DataFrame.
    lista_de_dataframes = [pd.read_csv(archivo) for archivo in lista_archivos_csv]
    df_consolidado = pd.concat(lista_de_dataframes, ignore_index=True)

    # --- GUARDADO EN NUEVA CARPETA  ---
    # 5. Define el nombre de la carpeta de salida.
    carpeta_salida = 'consolidados_nan'
    
    # 6. Crea la carpeta si no existe.
    #    exist_ok=True evita que salga un error si la carpeta ya fue creada.
    os.makedirs(carpeta_salida, exist_ok=True)

    # 7. Define el nombre del archivo de salida y crea la ruta completa.
    nombre_archivo_salida = 'datos_consolidados.csv'
    ruta_archivo_salida = os.path.join(carpeta_salida, nombre_archivo_salida)

    # 8. Guarda el DataFrame consolidado en la nueva ruta.
    df_consolidado.to_csv(ruta_archivo_salida, index=False)

    print(f"\n¡Proceso completado!")
    print(f"El resultado se ha guardado en: '{ruta_archivo_salida}'")

    # Muestra información del resultado.
    print("\nInformación general del DataFrame consolidado:")
    df_consolidado.info()
```

### REORDENAR EL CSV ' schema (columnas) ' 

```python
# 1. Define la ruta del archivo que ya creaste.
ruta_archivo_entrada = os.path.join('consolidados_nan', 'datos_consolidados.csv')

# 2. Lee el archivo CSV en un DataFrame.
try:
    df = pd.read_csv(ruta_archivo_entrada)
    
    # 3. (Paso Clave) Limpia los espacios en blanco de los nombres de las columnas.
    #    Esto convierte ' pm25' en 'pm25', por ejemplo.
    df.columns = df.columns.str.strip()

    # 4. Define el orden correcto y limpio de las columnas.
    columnas_ordenadas = [
        'estado', 
        'fuente_archivo', # Nombre corregido
        'date', 
        'pm25', 
        'pm10', 
        'o3', 
        'no2', 
        'so2', 
        'co' # Nombre corregido
    ]

    # 5. Reordena el DataFrame.
    df_reordenado = df[columnas_ordenadas]

    # 6. Define la ruta para el nuevo archivo y lo guarda.
    nombre_archivo_salida = 'datos_consolidados_ordenados.csv'
    ruta_archivo_salida = os.path.join('consolidados_nan', nombre_archivo_salida)
    
    df_reordenado.to_csv(ruta_archivo_salida, index=False)

    print(f"¡Reordenamiento completado!")
    print(f"El nuevo archivo se ha guardado en: '{ruta_archivo_salida}'")

    print("\nPrimeras 5 filas del archivo reordenado:")
    print(df_reordenado.head())

except FileNotFoundError:
    print(f"Error: No se encontró el archivo en la ruta '{ruta_archivo_entrada}'")
    print("Asegúrate de que el archivo existe en esa ubicación.")
```
